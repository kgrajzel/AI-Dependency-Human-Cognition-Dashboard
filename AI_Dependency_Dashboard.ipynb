# **Futuristic Data Exporing the Effects of AI on Human Cognition Dashboard Project 2**

# Overviews

In this project interactive dashboards were created to explor the AI Impact on Human Mental Growth dataset downloaded from kaggle.  

The dataset is a synthetic futuristic dataset designed to analyze how advanced artificial intelligence may influence human creativity, cognition, emotional intelligence, learning behavior, and decision-making across different societies and generations. The dataset simulates multiple future scenarios ranging from optimistic human-AI collaboration to pessimistic overdependence on AI systems. 

# Import Libraries

```python
import pandas as pd
import plotly.graph_objects as go
import plotly.express as px
from dash import Dash, dcc, html, Input, Output
```

# Loading the Data 

```python
df = pd.read_csv("AI_mental_growth.csv")
```

# Exploring the Data

```python
df.info()
df.describe()
for col in df.select_dtypes(include="object").columns:
    print(f"\n--- {col} ---")
    print(df[col].value_counts())
```

The data includes two time variables (Year and Ear), denoting the same information, one geograpgic variable (Region) and three demographic variables (age category denoted as Demographic Group, Education, and Professions). 

The data includes 18 continous (int/float) variables. The continous variables are Weekly_AI_Usage_Hours, AI_Dependency_Score, Human_Creativity_Score, Critical_Thinking_Ability, Problem_Solving_Skills, Emotional_Intelligence, Innovation_Rate, Learning_Speed, Human_Decision_Making_Involvement, Social_Interaction_Quality, Cognitive_Load_Index, Mental_Wellbeing_Score, AI_Tool_Proficiency, Adaptability_Score, Attention_Span_Minutes. 

The data also contains outcome variables such as Positive_Effects_of_AI, Negative_Effects_of_AI and Outcome_Label describing the ultimate result of AI on human cognition. 

# Variables Used in the Project 

The dashboard uses the following variables:

- **AI_Dependency_Score** — measure of AI dependency.
- **Profession** — professional category.
- **Scenario** — future scenario classification.
- **Year** — year associated with the observation.
- **Outcome_Label** — outcome classification provided in the dataset.
- **Cognitive variables** — continuous measures used to examine relationships
  with AI dependency.

# Dashboard Interactives

## Graph 1 - Cognitive Outcomes by Scenario

```python
@app.callback(
    Output("cognitive-scatter", "figure"),
    [
        Input("cognitive-variable-dropdown", "value"),
        Input("scenario-radio", "value")
    ]
)
def update_cognitive_scatter(selected_variable, selected_scenario):

    plot_df = df[
        df["Scenario"] == selected_scenario
    ].copy()

    scenario_colors = {
        "Optimistic": "#2E8B57",      
        "Neutral": "#3B82F6",         
        "Mixed": "#8B5CF6",           
        "Pessimistic": "#DC2626"     
    }

    fig = px.scatter(
        plot_df,
        x="AI_Dependency_Score",
        y=selected_variable,
        color="Scenario",
        hover_data=["Profession", "Region"],
        title=(
            f"AI Dependency vs. "
            f"{selected_variable.replace('_', ' ')}"
        ),
        color_discrete_map=scenario_colors
    )

    fig.update_layout(
        xaxis_title="AI Dependency Score",
        yaxis_title=selected_variable.replace("_", " "),
        legend_title="Scenario"
    )

    return fig
```

## Graph 2 - — Professions by Scenario

```python
@app.callback(
    Output("profession-heatmap", "figure"),
    Input("heatmap-scenario", "value")
)
def update_profession_heatmap(selected_scenario):

    # Mean AI dependency by profession and scenario
    summary = (
        df.groupby(["Profession", "Scenario"])["AI_Dependency_Score"]
          .mean()
          .reset_index()
    )

    scenario_data = (
        summary[summary["Scenario"] == selected_scenario]
        .sort_values("AI_Dependency_Score", ascending=False)
    )

    fig = go.Figure()

    fig.add_trace(
        go.Heatmap(
            z=[scenario_data["AI_Dependency_Score"]],
            x=scenario_data["Profession"],
            y=[selected_scenario],
            text=[scenario_data["AI_Dependency_Score"].round(2)],
            texttemplate="%{text}",
            colorscale="Viridis",
            colorbar=dict(
                title="AI Dependency"
            )
        )
    )

    fig.update_layout(
        title=f"AI Dependency by Profession — {selected_scenario}",
        xaxis_title="Profession",
        yaxis_title="Scenario",
        xaxis_tickangle=-45,
        margin=dict(
            l=80,
            r=80,
            t=80,
            b=150
        )
    )

    return fig
```

## Graph 3 - Outcomes by Scenario and Year

```python
@app.callback(
    Output("outcome-scenario-year-graph", "figure"),
    Input("year-slider", "value")
)
def update_outcome_scenario_year(selected_year):

    # Filter to selected year
    plot_df = df[
        df["Year"] == selected_year].copy()

    # Count outcomes by scenario
    outcome_counts = (
        plot_df.groupby(
            ["Scenario", "Outcome_Label"]
        )
        .size()
        .reset_index(name="Count")
    )
    outcome_counts["Percentage"] = (
        outcome_counts["Count"] /
        outcome_counts.groupby("Scenario")["Count"]
        .transform("sum")
        * 100
    )

    # Outcome order
    outcome_order = [
        "Positive Growth",
        "Stagnant / Moderate",
        "Cognitive Decline Risk"
    ]

    outcome_colors = {
        "Positive Growth": "#2E8B57",
        "Stagnant / Moderate": "#3B82F6",
        "Cognitive Decline Risk": "#DC2626"
    }

    fig = px.bar(
        outcome_counts,
        x="Scenario",
        y="Percentage",
        color="Outcome_Label",
        text="Percentage",
        category_orders={
            "Outcome_Label": outcome_order
        },
        color_discrete_map=outcome_colors,
        title=f"Outcomes by Scenario — {selected_year}"
    )

    fig.update_traces(
        texttemplate="%{text:.1f}%",
        textposition="inside"
    )

    fig.update_layout(
        barmode="stack",
        yaxis=dict(
            title="Percentage of Outcomes",
            range=[0, 100]
        ),
        xaxis_title="Scenario",
        legend_title="Outcome",
        hovermode="x unified"
    )

    return fig
```
